### STEP 1: Load and Clean

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("loan_data.csv")
print(df.info())
```

```python
df.describe()
```

```python
# Handle missing values
print(df.isnull().sum())
df["income"] =df['income'].fillna(df['income'].median())
# summar statistics
print(df.info())
df.describe()
```

### STEP 2: Visualizing Correlations
 understanding how variables are related is key to risk assessment


```python
print("=== CORRELATION ANALYSIS ===")
# select numerical variables for correlation analysis
numerical_df = df[['loan_id','age', 'income', 'loan_amount', 'credit_score', 'default']]
corr_matrix = numerical_df.corr()

plt.figure(figsize=(10,6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, square=True, linewidth=0.5)
plt.title("Correlation Matrix of Numerical Values", fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```
